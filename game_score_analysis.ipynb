```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
scores = np.array([
    [55, 65, 75, 85, 95],
    [76, 78, 77, 79, 75],
    [40, 90, 45, 95, 80]
])

names = ["Nova", "Pixel", "Echo"]
```

```python
means = scores.mean(axis=1)
stds = scores.std(axis=1)

stats = np.column_stack((means, stds))
stats
```

```python
plt.bar(names, means)

plt.xlabel("Player Names")
plt.ylabel("Player Averages")
plt.title("Player Averages Across Rounds")
plt.ylim(0, 100)
plt.show()
```

```python
rounds = np.arange(1, 6)

plt.plot(rounds, scores[0], marker="o", label=names[0])
plt.plot(rounds, scores[1], marker="o", label=names[1])
plt.plot(rounds, scores[2], marker="o", label=names[2])

plt.xlabel("Round")
plt.ylabel("Score")
plt.title("Scores Across 5 Rounds")

plt.legend()
plt.show()
```

- Nova shows a steady incread across the 5 rounds
- Pixel stays relativley consistent throughout the rounds
- Echo has no consistentcy with high games and low games

(Ai rendition: Nova’s score increases by 10 points each round, from 55 to 95. Pixel stays between 75 and 79, making them the most consistent player. Echo’s scores fluctuate between 40 and 95; their average of 70 hides that large spread.)

! These five fictional rounds describe the observed scores but they don’t establish how the players will perform in future rounds !
